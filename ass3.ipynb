```python
from queue import PriorityQueue

graph = {}
heuristic = {}

n = int(input("Enter number of nodes: "))

for i in range(n):
    node = input("\nEnter node: ")
    neighbours = input("Enter neighbours separated by space: ").split()
    graph[node] = neighbours
    heuristic[node] = int(input("Enter heuristic value: "))

start = input("\nEnter Start Node: ")
goal = input("Enter Goal Node: ")

open_list = PriorityQueue()
open_list.put((heuristic[start], start))

closed_list = []
parent = {start: None}

step = 1
found = False

print("\n========== GREEDY BEST FIRST SEARCH ==========")

while not open_list.empty():

    # Display Open List
    temp = list(open_list.queue)
    temp.sort()

    print("\nStep", step)
    print("Open List:", [node for h, node in temp])
    print("Closed List:", closed_list)

    # Remove node with smallest heuristic
    h, current = open_list.get()

    if current in closed_list:
        continue

    # Add current node to Closed List
    closed_list.append(current)

    # Check goal
    if current == goal:
        found = True
        break

    # Add neighbours to Open List
    for neighbour in graph[current]:

        if neighbour not in closed_list:
            parent[neighbour] = current
            open_list.put((heuristic[neighbour], neighbour))

    step += 1


# Final Open and Closed Lists
temp = list(open_list.queue)
temp.sort()

print("\n========== FINAL RESULT ==========")
print("Open List:", [node for h, node in temp])
print("Closed List:", closed_list)


# Display Path
if found:
    path = []
    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    print("Final Path:", " -> ".join(path))
else:
    print("No Path Exists")
```
