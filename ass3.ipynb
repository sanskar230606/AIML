```python
from collections import defaultdict, deque

graph = defaultdict(list)
queue = deque()
path = []
nodes = {}

directed = int(input("Is the graph:\n1. Directed\n2. Undirected?\nChoice: "))
n = int(input("Enter number of nodes: "))

for i in range(n):
    node = input(f"Enter node {i+1}: ")
    nodes[node] = int(input(f"Enter heuristic value of {node}: "))
    if i == 0:
        queue.append(node)

e = int(input("Enter number of edges: "))
for i in range(e):
    first = input(f"Enter start node for edge {i+1}: ")
    second = input(f"Enter end node for edge {i+1}: ")
    graph[first].append(second)
    if directed == 2:
        graph[second].append(first)

goal = input("Enter goal node: ")
nodes[goal] = 0

def q_add(*args):
    global queue, path, nodes
    q_list = list(queue)
    for node in args:
        if node not in q_list and node not in path:
            q_list.append(node)

    q_list.sort(key=lambda x: nodes[x])
    queue = deque(q_list)

def check_node(path):
    return goal in path

def work_gbfs():
    global queue, path
    while queue:
        print("Queue:", list(queue))

        node = queue.popleft()
        path.append(node)
        print("Path:", path)

        if check_node(path):
            print(f"\nGoal '{node}' found!")
            return

        q_add(*graph[node])
    print("\nGoal not reachable.")

work_gbfs()
```
