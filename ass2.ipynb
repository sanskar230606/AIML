```python
from collections import deque, defaultdict

frontier = deque()
reached = []
graph = defaultdict(list)
found = False
value = ""


def get_graph():
    global graph, frontier, reached, found

    graph = defaultdict(list)
    frontier = deque()
    reached = []
    found = False

    n = int(input("Enter number of nodes: "))

    for i in range(n):
        parent = input(f"Node {i}: ")
        graph[parent] = []
        if i == 0:
            frontier.append(parent)

    c = int(input("Enter number of edges: "))

    for i in range(c):
        first = input(f"Enter first node for edge {i + 1}: ")
        second = input(f"Enter second node for edge {i + 1}: ")
        graph[first].append(second)
        graph[second].append(first)

    return graph


def work_BFS(graph):
    global found

    while found == False and len(frontier) != 0:
        node = frontier.popleft()
        reached.append(node)

        q_add(*graph[node])

        found = check_value(reached)

    if found == False:
        print(value, "NOT FOUND")


def work_DFS(graph):
    global found

    while found == False and len(frontier) != 0:
        node = frontier.pop()
        reached.append(node)

        q_add(*graph[node])

        found = check_value(reached)

    if found == False:
        print(value, "NOT FOUND")


def q_add(*args):
    for i in args:
        if i in frontier or i in reached:
            continue
        else:
            frontier.append(i)

    print("Queue:", list(frontier))
    print("Path:", reached)
    print()


def check_value(reached):
    if value in reached:
        print(value, "FOUND")
        return True

    return False


while True:

    choice = int(input("\n1. BFS \n2. DFS \n3. Exit \nEnter choice: "))

    match choice:

        case 1:
            graph = get_graph()

            value = input("Enter the final state to find: ")

            work_BFS(graph)

        case 2:
            graph = get_graph()

            value = input("Enter the final state to find: ")

            work_DFS(graph)

        case 3:
            print("Exited Code...")
            break

        case _:
            print("Invalid Input.")
```
